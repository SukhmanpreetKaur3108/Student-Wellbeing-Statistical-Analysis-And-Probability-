```python
# DAY 16 - STUDENT WELLBEING STATISTICAL ANALYSIS & PROBABILITY

# ================================================================
# 1. IMPORT LIBRARIES AND LOAD DATASET
# ================================================================

import pandas as pd
import numpy as np
from google.colab import files

# Select and upload the CSV file manually
uploaded = files.upload()

# Get the uploaded file name
file_name = list(uploaded.keys())[0]

df = pd.read_csv(file_name)

print("Dataset Shape:", df.shape)
print("\nFirst 5 Rows:")
display(df.head())

print("\nColumn Names:")
print(df.columns.tolist())

print("\nData Types:")
print(df.dtypes)

print("\nMissing Values:")
display(df.isnull().sum())

print("\nDuplicate Rows:", df.duplicated().sum())


# ================================================================
# 2. DESCRIPTIVE STATISTICS
# ================================================================

variables = [
    "Weekly_Study_Hours",
    "Average_Sleep_Hours",
    "Daily_Screen_Time_Hours",
    "Stress_Score",
    "Academic_Readiness_Score"
]

print("\n" + "=" * 70)
print("DESCRIPTIVE STATISTICS")
print("=" * 70)

statistics = []

for column in variables:
    statistics.append({
        "Variable": column,
        "Mean": df[column].mean(),
        "Median": df[column].median(),
        "Mode": df[column].mode().iloc[0],
        "Range": df[column].max() - df[column].min(),
        "Variance": df[column].var(),
        "Standard Deviation": df[column].std(),
        "Q1": df[column].quantile(0.25),
        "Q3": df[column].quantile(0.75),
        "IQR": df[column].quantile(0.75) - df[column].quantile(0.25)
    })

stats_df = pd.DataFrame(statistics)

display(stats_df.round(2))


# ================================================================
# 3. FORMULAS AND INTERPRETATION
# ================================================================

print("""
FORMULAS USED

Mean = Sum of all observations / Number of observations

Median = Middle value after arranging the observations

Mode = Most frequently occurring value

Range = Maximum value - Minimum value

Variance = Average squared deviation from the mean

Standard Deviation = Square root of variance

Q1 = 25th percentile
Q3 = 75th percentile

IQR = Q3 - Q1
""")

# Variable with greatest variability based on standard deviation
greatest_variability = stats_df.loc[
    stats_df["Standard Deviation"].idxmax(), "Variable"
]

print("Variable with greatest variability based on Standard Deviation:",
      greatest_variability)

print("""
INTERPRETATION:
The variable with the largest standard deviation shows the greatest
overall variability around its mean.
""")


# ================================================================
# 4. IQR OUTLIER DETECTION
# ================================================================

outlier_variables = [
    "Weekly_Study_Hours",
    "Daily_Screen_Time_Hours",
    "Commute_Time_Minutes",
    "Monthly_Discretionary_Spending"
]

print("\n" + "=" * 70)
print("IQR OUTLIER ANALYSIS")
print("=" * 70)

outlier_summary = []

for column in outlier_variables:

    q1 = df[column].quantile(0.25)
    q3 = df[column].quantile(0.75)
    iqr = q3 - q1

    lower_limit = q1 - 1.5 * iqr
    upper_limit = q3 + 1.5 * iqr

    outliers = df[
        (df[column] < lower_limit) |
        (df[column] > upper_limit)
    ]

    outlier_summary.append({
        "Variable": column,
        "Q1": q1,
        "Q3": q3,
        "IQR": iqr,
        "Lower Limit": lower_limit,
        "Upper Limit": upper_limit,
        "Number of Outliers": len(outliers)
    })

outlier_df = pd.DataFrame(outlier_summary)

display(outlier_df.round(2))

print("""
IQR OUTLIER FORMULA

Lower Bound = Q1 - 1.5 × IQR
Upper Bound = Q3 + 1.5 × IQR

Any observation below the lower bound or above the upper bound
is considered a potential outlier.
""")


# ================================================================
# 5. BEFORE AND AFTER OUTLIER REMOVAL
#    Using Weekly_Study_Hours
# ================================================================

column = "Weekly_Study_Hours"

q1 = df[column].quantile(0.25)
q3 = df[column].quantile(0.75)
iqr = q3 - q1

lower_limit = q1 - 1.5 * iqr
upper_limit = q3 + 1.5 * iqr

clean_study_hours = df[
    (df[column] >= lower_limit) &
    (df[column] <= upper_limit)
][column]

before_mean = df[column].mean()
before_median = df[column].median()

after_mean = clean_study_hours.mean()
after_median = clean_study_hours.median()

comparison = pd.DataFrame({
    "Measure": ["Mean", "Median"],
    "Before Outlier Removal": [before_mean, before_median],
    "After Outlier Removal": [after_mean, after_median]
})

print("\n" + "=" * 70)
print("WEEKLY STUDY HOURS: BEFORE VS AFTER OUTLIER REMOVAL")
print("=" * 70)

display(comparison.round(2))

print("""
INTERPRETATION:
The mean is generally more sensitive to extreme values than the median.
Therefore, removing outliers can change the mean more noticeably.
The comparison above shows the actual effect of detected study-hour
outliers on this dataset.
""")


# ================================================================
# 6. DEFINE PROBABILITY EVENTS
# ================================================================

# Event A: Part-time job = Yes
A = df["Part_Time_Job"] == "Yes"

# Event B: Stress Score >= 7
B = df["Stress_Score"] >= 7

# Event C: Scholarship = Yes
C = df["Scholarship"] == "Yes"

# Event D: Exercises at least 3 days per week
D = df["Exercise_Days_Per_Week"] >= 3

total_students = len(df)

print("\n" + "=" * 70)
print("PROBABILITY EVENTS")
print("=" * 70)

print("Total students:", total_students)
print("Event A - Part-time job:", A.sum())
print("Event B - Stress Score >= 7:", B.sum())
print("Event C - Scholarship:", C.sum())
print("Event D - Exercise >= 3 days/week:", D.sum())


# ================================================================
# 7. BASIC PROBABILITIES
# ================================================================

P_A = A.mean()
P_B = B.mean()
P_C = C.mean()
P_D = D.mean()

print("\n" + "=" * 70)
print("BASIC PROBABILITIES")
print("=" * 70)

print("Formula: P(Event) = Number of favorable outcomes / Total outcomes")

print(f"\nP(A) = {A.sum()} / {total_students} = {P_A:.4f}")
print(f"P(B) = {B.sum()} / {total_students} = {P_B:.4f}")
print(f"P(C) = {C.sum()} / {total_students} = {P_C:.4f}")
print(f"P(D) = {D.sum()} / {total_students} = {P_D:.4f}")

print("""
INTERPRETATION:
P(A) represents the probability that a randomly selected student
has a part-time job.

P(B) represents the probability that a randomly selected student
has a Stress Score of at least 7.

P(C) represents the probability that a randomly selected student
has a scholarship.

P(D) represents the probability that a randomly selected student
exercises at least 3 days per week.
""")


# ================================================================
# 8. UNION, INTERSECTION AND CONDITIONAL PROBABILITIES
# ================================================================

A_or_B = A | B
A_and_B = A & B

P_A_or_B = A_or_B.mean()
P_A_and_B = A_and_B.mean()

P_A_given_B = A_and_B.sum() / B.sum()
P_B_given_A = A_and_B.sum() / A.sum()

print("\n" + "=" * 70)
print("COMBINED AND CONDITIONAL PROBABILITIES")
print("=" * 70)

print("""
Formulas:

P(A or B) = P(A) + P(B) - P(A and B)

P(A and B) = Number of students satisfying both A and B / Total

P(A|B) = P(A and B) / P(B)

P(B|A) = P(A and B) / P(A)
""")

print(f"P(A or B)  = {P_A_or_B:.4f}")
print(f"P(A and B) = {P_A_and_B:.4f}")
print(f"P(A|B)     = {P_A_given_B:.4f}")
print(f"P(B|A)     = {P_B_given_A:.4f}")

print("""
INTERPRETATION:
P(A or B) represents the probability that a student either has a
part-time job or has a Stress Score of at least 7, including students
who satisfy both conditions.

P(A and B) represents the probability that both conditions occur.

P(A|B) represents the probability of having a part-time job given
that the student has a Stress Score of at least 7.

P(B|A) represents the probability of having a Stress Score of at least
7 given that the student has a part-time job.
""")


# ================================================================
# 9. MUTUALLY EXCLUSIVE EVENTS
# ================================================================

year_1 = df["Year_of_Study"] == 1
year_4 = df["Year_of_Study"] == 4

intersection_years = (year_1 & year_4).sum()

print("\n" + "=" * 70)
print("MUTUALLY EXCLUSIVE EVENTS")
print("=" * 70)

print("Students in Year 1:", year_1.sum())
print("Students in Year 4:", year_4.sum())
print("P(Year 1 and Year 4):", intersection_years / total_students)

if intersection_years == 0:
    print("\nConclusion: Year 1 and Year 4 are mutually exclusive events.")
else:
    print("\nConclusion: Year 1 and Year 4 are not mutually exclusive.")

print("""
Reason:
A student cannot simultaneously belong to Year 1 and Year 4.
Therefore, the intersection of these two events is zero.
""")


# ================================================================
# 10. TEST WHETHER A AND B APPEAR INDEPENDENT
# ================================================================

independent_product = P_A * P_B

print("\n" + "=" * 70)
print("INDEPENDENCE OF EVENTS A AND B")
print("=" * 70)

print(f"P(A and B)     = {P_A_and_B:.4f}")
print(f"P(A) × P(B)    = {independent_product:.4f}")
print(f"Difference     = {abs(P_A_and_B - independent_product):.4f}")

print("""
Formula for independence:

P(A and B) = P(A) × P(B)

If the two values are approximately equal, A and B appear independent.
If they differ substantially, A and B appear dependent.
""")

if np.isclose(P_A_and_B, independent_product, atol=0.01):
    print("Conclusion: A and B appear approximately independent.")
else:
    print("Conclusion: A and B do not appear independent; they show evidence of dependence.")


# ================================================================
# 11. BAYES' THEOREM
# ================================================================

# P(B|A)
P_B_given_A = P_B_given_A

# P(B|not A)
not_A = ~A
P_B_given_not_A = B[not_A].mean()

# Bayes' theorem
P_A_given_B_bayes = (
    P_B_given_A * P_A
) / (
    P_B_given_A * P_A +
    P_B_given_not_A * (1 - P_A)
)

print("\n" + "=" * 70)
print("BAYES' THEOREM")
print("=" * 70)

print("""
Bayes' Theorem:

P(A|B) =
[P(B|A) × P(A)]
-------------------------------
[P(B|A) × P(A)] + [P(B|not A) × P(not A)]
""")

print(f"P(B|A)       = {P_B_given_A:.4f}")
print(f"P(B|not A)   = {P_B_given_not_A:.4f}")
print(f"P(A)         = {P_A:.4f}")
print(f"P(not A)     = {1 - P_A:.4f}")

print(f"\nBayes P(A|B) = {P_A_given_B_bayes:.4f}")
print(f"Direct P(A|B) = {P_A_given_B:.4f}")
print(f"Difference    = {abs(P_A_given_B_bayes - P_A_given_B):.10f}")

if np.isclose(P_A_given_B_bayes, P_A_given_B):
    print("\nVerification: Bayes' theorem agrees with the direct calculation.")
else:
    print("\nVerification: The values differ; check the event calculations.")

print("""
INTERPRETATION:
Bayes' theorem gives the probability of Event A after considering
information about Event B. The result should agree with the direct
conditional probability P(A|B), apart from negligible rounding
differences.
""")


# ================================================================
# 12. NORMAL DISTRIBUTION - ACADEMIC READINESS SCORE
# ================================================================

readiness = df["Academic_Readiness_Score"]

mean_readiness = readiness.mean()
std_readiness = readiness.std()

highest_score = readiness.max()
lowest_score = readiness.min()

highest_z = (highest_score - mean_readiness) / std_readiness
lowest_z = (lowest_score - mean_readiness) / std_readiness

print("\n" + "=" * 70)
print("NORMAL DISTRIBUTION - ACADEMIC READINESS SCORE")
print("=" * 70)

print("""
Z-score formula:

Z = (X - Mean) / Standard Deviation
""")

print(f"Mean Academic Readiness Score = {mean_readiness:.2f}")
print(f"Standard Deviation             = {std_readiness:.2f}")

print(f"\nHighest Academic Readiness Score = {highest_score:.2f}")
print(f"Z-score of highest score         = {highest_z:.2f}")

print(f"\nLowest Academic Readiness Score = {lowest_score:.2f}")
print(f"Z-score of lowest score        = {lowest_z:.2f}")

print("""
INTERPRETATION:

The Z-score of the highest-scoring student tells how many standard
deviations that student's score is above the mean.

The Z-score of the lowest-scoring student tells how many standard
deviations that student's score is below the mean.

A larger absolute Z-score indicates that the observation is farther
from the average.
""")


# ================================================================
# 13. 68-95-99.7 EMPIRICAL RULE
# ================================================================

print("\n" + "=" * 70)
print("68-95-99.7 EMPIRICAL RULE")
print("=" * 70)

print("""
For an approximately normal distribution:

Within ±1 standard deviation → approximately 68%
Within ±2 standard deviations → approximately 95%
Within ±3 standard deviations → approximately 99.7%
""")

for sd in [1, 2, 3]:

    lower = mean_readiness - sd * std_readiness
    upper = mean_readiness + sd * std_readiness

    percentage = ((readiness >= lower) & (readiness <= upper)).mean() * 100

    expected = {1: 68, 2: 95, 3: 99.7}[sd]

    print(f"\nWithin ±{sd} SD:")
    print(f"Score range       = {lower:.2f} to {upper:.2f}")
    print(f"Actual percentage = {percentage:.2f}%")
    print(f"Expected percentage by empirical rule = {expected}%")


# ================================================================
# 14. FINAL FIVE STATISTICAL OBSERVATIONS
# ================================================================

print("\n" + "=" * 70)
print("FIVE MEANINGFUL STATISTICAL OBSERVATIONS")
print("=" * 70)

# Get useful values for the observations
highest_sd_row = stats_df.loc[stats_df["Standard Deviation"].idxmax()]

most_outliers_row = outlier_df.loc[
    outlier_df["Number of Outliers"].idxmax()
]

print(f"""
1. Among the five main numerical variables,
   {greatest_variability} has the greatest variability based on
   standard deviation ({highest_sd_row["Standard Deviation"]:.2f}).

2. The IQR method identified {most_outliers_row["Number of Outliers"]}
   potential outliers in {most_outliers_row["Variable"]}, which has
   the largest number of detected outliers among the four variables
   examined.

3. Weekly study hours changed from a mean of {before_mean:.2f}
   before outlier removal to {after_mean:.2f} after removing the
   detected outliers, showing the effect extreme observations can
   have on the mean.

4. The probability analysis gives P(A) = {P_A:.4f} and
   P(B) = {P_B:.4f}. The comparison of P(A and B) = {P_A_and_B:.4f}
   with P(A) × P(B) = {independent_product:.4f} indicates that
   A and B {'appear approximately independent' if np.isclose(P_A_and_B, independent_product, atol=0.01) else 'do not appear independent'}.

5. Academic Readiness Score has a mean of {mean_readiness:.2f}
   and a standard deviation of {std_readiness:.2f}. The Z-scores of
   the highest and lowest scoring students show how far these extreme
   observations are from the average student.
""")

print("=" * 70)
print("ANALYSIS COMPLETED")
print("=" * 70)
```
